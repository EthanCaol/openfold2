```python
import torch


def empty_slicer(t):
    return [slice(None) for _ in t.shape]


def slice_tensor(t, start, end, dim):
    s = empty_slicer(t)
    s[dim] = slice(start, end)
    return t[tuple(s)]


t = torch.arange(24).reshape(2, 3, 4)
print(t)
print(slice_tensor(t, 1, 3, 1))
```

```python
t = torch.arange(12).reshape(3,4)
row_idx = [0,1]
col_idx = [2,3]
print(t)
print(t[row_idx, col_idx])
print(t[0:1, 2:3])
```
