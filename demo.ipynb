```python
import torch

# 创建示例张量
x = torch.tensor([
    [[1, 2], [3, 4]],   # 第0个3D切片
    [[5, 6], [7, 8]],   # 第1个3D切片
    [[9, 10], [11, 12]] # 第2个3D切片
])
# x.shape = [3, 2, 2]

print("原始张量：")
print(x)
print(f"形状: {x.shape}")

torch.set_float32_matmul_precision("medium")
print(torch.get_float32_matmul_precision())
```

```python
# 选择具体的元素
print("\n1. 多维索引 x[0, 1, 0]:")
print(x[0, 1, 0])  # 输出：tensor(3)
# 选择：第0个3D切片 → [[1,2],[3,4]]
#       第1个2D行 → [3,4]
#       第0个元素 → 3

print("\n2. 用列表索引（当前有警告，未来会改变）:")
slices = [0, 1, 0]
print(f"当前: {x[slices]}")           # 现在：输出 3
print(f"未来: {x[tuple(slices)]}")  # 应该用这个
```

```python
print("\n3. 张量索引 - 选择多个3D切片:")
indices = torch.tensor([0, 2])  # 选择第0和第2个3D切片
result = x[indices]
print(f"x[[0, 2]] 的形状: {result.shape}")  # torch.Size([2, 2, 2])
print(f"结果:\n{result}")
# 输出：
# [[[1, 2], [3, 4]],  # 第0个切片
#  [[9, 10], [11, 12]]] # 第2个切片
```
